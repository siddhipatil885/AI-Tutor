# Phase 1: Dataset Exploration

This notebook explores the McMiner dataset provided in the repository. We aim to:
- Locate the dataset files
- Inspect the data format, shape, and fields
- Analyze misconception label distribution and data quality

```python
import os
import pandas as pd
import glob

# Define the project root and expected dataset paths
PROJECT_ROOT = os.path.abspath('..')
print(f"Project Root: {PROJECT_ROOT}")
```

## 1. Locate the Dataset
The user stated the McMiner dataset is in the repository. We will dynamically search for files.

```python
def find_potential_datasets(root_dir):
    dataset_files = []
    for dirpath, _, filenames in os.walk(root_dir):
        if '.git' in dirpath or 'node_modules' in dirpath:
            continue
        for f in filenames:
            if f.endswith('.csv') or f.endswith('.json') or f.endswith('.jsonl') or f.endswith('.parquet'):
                dataset_files.append(os.path.join(dirpath, f))
    return dataset_files

all_files = find_potential_datasets(PROJECT_ROOT)
print(f"Found {len(all_files)} potential data files in the repository.")
```

## 2. Inspecting the `Project_CodeNet-main` folder
A directory named `Project_CodeNet-main` was found in the project root. Project CodeNet is the dataset McMiner is built upon. Let's inspect what is inside this directory.

```python
codenet_dir = os.path.join(PROJECT_ROOT, 'Project_CodeNet-main')
if os.path.exists(codenet_dir):
    print("Found Project_CodeNet-main directory.")
    print("Contents:")
    for item in os.listdir(os.path.join(codenet_dir, 'Project_CodeNet-main')):
        print(" -", item)
else:
    print("Project_CodeNet-main not found.")
```
